# Understanding Digital Images Using Python

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
filepath = "images/image_1.jpg"
image = cv2.imread(filepath)
if image is None:
    raise FileNotFoundError("Image could not be read.")

plt.figure(figsize=(7, 5))
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Picture")
plt.axis("off")
plt.show()
```

# Threshold Transformation

```python
# Change Gray Image

def transform(t: int, image: np.ndarray):
    _, binary_image = cv2.threshold(image, t, 255, cv2.THRESH_BINARY)
    return binary_image


_gray_image = copy.deepcopy(gray_image)

plt.figure(figsize=(10, 14))

plt.subplot(3, 2, 1)
plt.imshow(_gray_image, cmap="gray")
plt.axis("off")
plt.title('Original')

gray_image_50 = transform(t=50, image=_gray_image)
plt.subplot(3, 2, 2)
plt.imshow(gray_image_50, cmap="gray")
plt.axis("off")
plt.title('T=50')

gray_image_100 = transform(t=100, image=_gray_image)
plt.subplot(3, 2, 3)
plt.imshow(gray_image_100, cmap="gray")
plt.axis("off")
plt.title('T=100')

gray_image_180 = transform(t=180, image=_gray_image)
plt.subplot(3, 2, 4)
plt.imshow(gray_image_180, cmap="gray")
plt.axis("off")
plt.title('T=180')

gray_image_220 = transform(t=220, image=_gray_image)
plt.subplot(3, 2, 5)
plt.imshow(gray_image_220, cmap="gray")
plt.axis("off")
plt.title('T=220')

plt.show()
```

# Salt and Pepper Noise Filtering

```python
def add_salt_and_pepper_noise(image, noise_ratio: float = 0.05, salt_ratio: float = 0.5):
    noisy_image = image.copy()

    h, w = noisy_image.shape[:2]
    rand_matrix = np.random.rand(h, w)

    pepper_threshold = noise_ratio * (1 - salt_ratio)
    salt_threshold = noise_ratio

    pepper_mask = rand_matrix < pepper_threshold
    noisy_image[pepper_mask] = 0

    salt_mask = (rand_matrix >= pepper_threshold) & (rand_matrix < salt_threshold)
    noisy_image[salt_mask] = 255

    return noisy_image
```

```python
# Adding salt and pepper noise to a grayscale image

noisy_gray = add_salt_and_pepper_noise(gray_image, noise_ratio=0.05)

fig, axes = plt.subplots(2, 2, figsize=(14, 8))

# (A) Original Image
axes[0, 0].imshow(gray_image, cmap='gray')
axes[0, 0].set_title('Original Grayscale Image', fontsize=12)
axes[0, 0].axis('off')

# (B) Noisy Image
axes[0, 1].imshow(noisy_gray, cmap='gray')
axes[0, 1].set_title('Noisy Image (5% Salt-and-Pepper)', fontsize=12)
axes[0, 1].axis('off')

# (C) Original Histogram
hist_orig = cv2.calcHist([gray_image], [0], None, [256], [0, 256])
axes[1, 0].plot(hist_orig, color='black')
axes[1, 0].set_title('Original Histogram')
axes[1, 0].set_xlim([0, 256])
axes[1, 0].set_xlabel('Pixel Intensity')
axes[1, 0].set_ylabel('Count')
axes[1, 0].grid(True, linestyle='--', alpha=0.5)

# (D) Noisy Histogram (Notice sharp spikes at bin 0 and bin 255)
hist_noisy = cv2.calcHist([noisy_gray], [0], None, [256], [0, 256])
axes[1, 1].plot(hist_noisy, color='crimson')
axes[1, 1].set_title('Noisy Histogram (Spikes at 0 & 255)')
axes[1, 1].set_xlim([0, 256])
axes[1, 1].set_xlabel('Pixel Intensity')
axes[1, 1].set_ylabel('Count')
axes[1, 1].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
# Adding salt and pepper noise to a RGB image

noisy_image = add_salt_and_pepper_noise(image, noise_ratio=0.05)

fig, axes = plt.subplots(2, 2, figsize=(14, 8))

# (A) Original Image
axes[0, 0].imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
axes[0, 0].set_title('Original RGB Image', fontsize=12)
axes[0, 0].axis('off')

# (B) Noisy Image
axes[0, 1].imshow(cv2.cvtColor(noisy_image, cv2.COLOR_BGR2RGB))
axes[0, 1].set_title('Noisy Image (5% Salt-and-Pepper)', fontsize=12)
axes[0, 1].axis('off')

# (C) Original Histogram (Plotting R, G, B channels)
colors = ('red', 'green', 'blue')
for i, col in enumerate(colors):
    hist_orig = cv2.calcHist([image], [i], None, [256], [0, 256])
    axes[1, 0].plot(hist_orig, color=col, label=f'{col.capitalize()}')
axes[1, 0].set_title('Original RGB Histogram')
axes[1, 0].set_xlim([0, 256])
axes[1, 0].set_xlabel('Pixel Intensity (0-255)')
axes[1, 0].set_ylabel('Count')
axes[1, 0].legend()
axes[1, 0].grid(True, linestyle='--', alpha=0.5)

# (D) Noisy Histogram (Shows identical equal spikes at 0 and 255 across all channels)
for i, col in enumerate(colors):
    hist_noisy = cv2.calcHist([noisy_image], [i], None, [256], [0, 256])
    axes[1, 1].plot(hist_noisy, color=col, label=f'{col.capitalize()}')
axes[1, 1].set_title('Noisy RGB Histogram (Equal Spikes at 0 & 255)')
axes[1, 1].set_xlim([0, 256])
axes[1, 1].set_xlabel('Pixel Intensity (0-255)')
axes[1, 1].set_ylabel('Count')
axes[1, 1].legend()
axes[1, 1].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

#  Filtering Using Mean and Median

```python
# Apply Linear Filter(Mean)
mean_filtered = cv2.blur(noisy_image, ksize=(3, 3))

# Apply Non-Linear Order Statistic(Median)
median_filtered = cv2.medianBlur(noisy_image, ksize=3)

fig, axes = plt.subplots(2, 2, figsize=(14, 12))

# (A) Original Image
axes[0, 0].imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
axes[0, 0].set_title('1. Original Clean Image', fontsize=12, fontweight='bold')
axes[0, 0].axis('off')
# (B) Noisy Image
axes[0, 1].imshow(cv2.cvtColor(noisy_image, cv2.COLOR_BGR2RGB))
axes[0, 1].set_title('2. Noisy Image (5% Salt-and-Pepper)', fontsize=12, fontweight='bold')
axes[0, 1].axis('off')

# (C) Mean Filter (Linear)
axes[1, 0].imshow(cv2.cvtColor(mean_filtered, cv2.COLOR_BGR2RGB))
axes[1, 0].set_title('3. Mean Filter (3x3) - Blurry Smudges', fontsize=12, fontweight='bold', color='darkred')
axes[1, 0].axis('off')
# (D) Median Filter (Order-Statistic)
axes[1, 1].imshow(cv2.cvtColor(median_filtered, cv2.COLOR_BGR2RGB))
axes[1, 1].set_title('4. Median Filter (3x3) - Clean Restoration', fontsize=12, fontweight='bold', color='darkgreen')
axes[1, 1].axis('off')

plt.tight_layout()
plt.show()
```

# Manual Statistic Calculation

```python
# 1. Define the 3x3 neighborhood matrix
neighborhood = np.array([
    [42,  43,  44],
    [43, 255,  45],
    [44,  45,  46]
])

# 2. Compute the 5 statistics using NumPy
val_min      = np.min(neighborhood)
val_max      = np.max(neighborhood)
val_median   = np.median(neighborhood)
val_mean     = np.mean(neighborhood)
val_midpoint = (val_min + val_max) / 2.0

# 3. Print the results formatted cleanly
print("--- Order-Statistic Exercise Results ---")
print(f"Sorted Values : {np.sort(neighborhood.ravel())}")
print(f"Minimum       : {val_min}")
print(f"Maximum       : {val_max}")
print(f"Median        : {val_median}")
print(f"Mean          : {val_mean:.2f}")
print(f"Midpoint      : {val_midpoint:.2f}")
```
